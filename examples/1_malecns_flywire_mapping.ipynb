# Male CNS <-> FlyWire mapping 

In this notebook we will demonstrate how to generate a mapping between the male CNS dataset (https://neuprint.janelia.org/) and FlyWire (https://codex.flywire.ai/).

```python
import cocoa as cc
```

First initialize the datasets - this lets you set certain parameters such as the datasource, version and so on.

Here, we will simply use the defaults:

```python
fw = cc.FlyWire()
mcns = cc.MaleCNS()
```

Now we can ask `cocoa` to generate a mapping between neurons in the two datasets.

We will use the `GraphMapper` class for this. It uses a graph-based approach where each neuron is associated with a set of labels.
For the male CNS, for example, we can use the `type`, `flywireType`, `hemibrainType` and `mancType`. The `GraphMapper` will take 
these labels and try to find the most granular mapping from on dataset to the other:

<div>
<center>
<img src="_static/graphmapper_illustration.png" width="600"/>
</center>
</div>

```python
# Intialize the mapper and compute the mapping
mapper = cc.GraphMapper(
    fw, mcns
).compile()
```

The mapper contains our mappings plus a few convenience functions to explore them:

```python
# The raw mappings are stored in a dictionary: {"dataset:id": "label"}
mapper.mappings_['FlyWire:720575940611299481']
```

All neurons with the same `label` belong to the same cross-matched group. The label itself is somewhat arbitrarily choosen from the 
pool of cell types in that group.

```python
# We can also get the inverse: {label: ["dataset:id", ...]}
mapper.mappings_inv_['CB2088']
```

```python
# Want to strip the {dataset} prefixes? No problem:
mapper.get_mappings(strip_prefix=True)[720575940611299481]
```

```python
# Each label is a connected component. We can figure out what is is composed of:
G = mapper.get_subgraph('CB2088', full_graph=True)

# Save to GML and load in e.g. cytoscape
# import networkx as nx
# nx.write_gml(G, "cb2088.gml")

G.nodes
```
